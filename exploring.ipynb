```python
import json

from databricks.sdk import WorkspaceClient

SPACE_ID = "01f1a09b07691e458088dc367cd0e81b"
NEW_TABLE = "propel_data.bronze.propel_othername_pfile_20050523_20260111"

w = WorkspaceClient()
space = w.genie.get_space(
    space_id=SPACE_ID,
    include_serialized_space=True,
)

if not space.serialized_space:
    raise RuntimeError("Genie Space does not contain serialized_space.")

table_info = w.tables.get(NEW_TABLE)
column_configs = [
    {
        "column_name": column.name,
        "enable_format_assistance": True,
    }
    for column in table_info.columns or []
]
column_configs.sort(key=lambda column: column["column_name"])

serialized = json.loads(space.serialized_space)
tables = serialized.setdefault("data_sources", {}).setdefault("tables", [])

if any(table.get("identifier") == NEW_TABLE for table in tables):
    print(f"Table already exists: {NEW_TABLE}")
else:
    tables.append(
        {
            "identifier": NEW_TABLE,
            "column_configs": column_configs,
        }
    )
    tables.sort(key=lambda table: table["identifier"])

    updated_serialized_space = json.dumps(serialized, indent=2)
    if space.serialized_space.endswith("\n"):
        updated_serialized_space += "\n"

    updated_space = w.genie.update_space(
        space_id=SPACE_ID,
        serialized_space=updated_serialized_space,
    )

    print(f"Added table: {NEW_TABLE}")
    print(f"Updated space: {updated_space.title}")

space_after = w.genie.get_space(
    space_id=SPACE_ID,
    include_serialized_space=True,
)

if not space_after.serialized_space:
    raise RuntimeError("Updated Genie Space does not contain serialized_space.")

final_tables = json.loads(space_after.serialized_space).get("data_sources", {}).get("tables", [])

if not any(table.get("identifier") == NEW_TABLE for table in final_tables):
    raise RuntimeError(f"Table was not found after update: {NEW_TABLE}")

print(f"Verified table: {NEW_TABLE}")
```

```python
import json

from databricks.sdk import WorkspaceClient

SPACE_ID = "01f1a09b07691e458088dc367cd0e81b"
TABLE_TO_REMOVE = "propel_data.bronze.propel_othername_pfile_20050523_20260111"

w = WorkspaceClient()
space = w.genie.get_space(
    space_id=SPACE_ID,
    include_serialized_space=True,
)

if not space.serialized_space:
    raise RuntimeError("Genie Space does not contain serialized_space.")

serialized = json.loads(space.serialized_space)
tables = serialized.setdefault("data_sources", {}).setdefault("tables", [])
updated_tables = [
    table
    for table in tables
    if table.get("identifier") != TABLE_TO_REMOVE
]

if len(updated_tables) == len(tables):
    print(f"Table is not present: {TABLE_TO_REMOVE}")
else:
    updated_tables.sort(key=lambda table: table["identifier"])
    serialized["data_sources"]["tables"] = updated_tables

    updated_serialized_space = json.dumps(serialized, indent=2)
    if space.serialized_space.endswith("\n"):
        updated_serialized_space += "\n"

    updated_space = w.genie.update_space(
        space_id=SPACE_ID,
        serialized_space=updated_serialized_space,
    )

    print(f"Removed table: {TABLE_TO_REMOVE}")
    print(f"Updated space: {updated_space.title}")

space_after = w.genie.get_space(
    space_id=SPACE_ID,
    include_serialized_space=True,
)

if not space_after.serialized_space:
    raise RuntimeError("Updated Genie Space does not contain serialized_space.")

final_tables = json.loads(space_after.serialized_space).get("data_sources", {}).get("tables", [])

if any(table.get("identifier") == TABLE_TO_REMOVE for table in final_tables):
    raise RuntimeError(f"Table was not removed: {TABLE_TO_REMOVE}")

print(f"Verified removal: {TABLE_TO_REMOVE}")
```

```python
import json

from databricks.sdk import WorkspaceClient

SPACE_ID = "01f1a09b07691e458088dc367cd0e81b"

TEXT_INSTRUCTION = """You are a Databricks SQL query generation assistant. For every request, return only one valid Databricks SQL query that answers the request. Do not execute the query. Do not return query results, a natural-language answer, markdown, explanations, comments, or code fences. Use only tables and columns available in this Genie space. Always use fully qualified table names in the form catalog.schema.table. When a request is ambiguous, choose the most reasonable interpretation using the available schema; do not invent tables, columns, values, joins, filters, or business rules."""

EXAMPLES_TO_ADD = [
    {
        "id": "f1160c6f4b9d4634a80593fe7d2c0021",
        "question": ["Show me a sample of the bronze provider location table"],
        "sql": [
            "SELECT *\n"
            "FROM propel_data.bronze.propel_pl_pfile_20050523_20260111\n"
            "LIMIT 5"
        ],
        "usage_guidance": [
            "Use this example when a user asks to inspect bronze-layer provider location data."
        ],
    },
    {
        "id": "f1160c6f4b9d4634a80593fe7d2c0022",
        "question": ["Show me a sample of the gold provider location table"],
        "sql": [
            "SELECT *\n"
            "FROM propel_data.gold.propel_pl_pfile_20050523_20260111\n"
            "LIMIT 5"
        ],
        "usage_guidance": [
            "Use this example when a user asks to inspect gold-layer provider location data."
        ],
    },
]

w = WorkspaceClient()
space = w.genie.get_space(
    space_id=SPACE_ID,
    include_serialized_space=True,
)

if not space.serialized_space:
    raise RuntimeError("Genie Space does not contain serialized_space.")

serialized = json.loads(space.serialized_space)
instructions = serialized.setdefault("instructions", {})
text_instructions = instructions.setdefault("text_instructions", [])

if text_instructions:
    text_instructions[0]["content"] = [TEXT_INSTRUCTION]
else:
    text_instructions.append(
        {
            "id": "f1160c6f4b9d4634a80593fe7d2c0020",
            "content": [TEXT_INSTRUCTION],
        }
    )

example_question_sqls = instructions.setdefault("example_question_sqls", [])
example_ids = {example.get("id") for example in example_question_sqls}
example_question_sqls.extend(
    example for example in EXAMPLES_TO_ADD if example["id"] not in example_ids
)

updated_serialized_space = json.dumps(serialized, indent=2)
if space.serialized_space.endswith("\n"):
    updated_serialized_space += "\n"

updated_space = w.genie.update_space(
    space_id=SPACE_ID,
    serialized_space=updated_serialized_space,
)

space_after = w.genie.get_space(
    space_id=SPACE_ID,
    include_serialized_space=True,
)

if not space_after.serialized_space:
    raise RuntimeError("Updated Genie Space does not contain serialized_space.")

updated_instructions = json.loads(space_after.serialized_space)["instructions"]
updated_examples = updated_instructions.get("example_question_sqls", [])
updated_example_ids = {example.get("id") for example in updated_examples}

if updated_instructions["text_instructions"][0].get("content") != [TEXT_INSTRUCTION]:
    raise RuntimeError("Text instructions were not updated.")

if not {example["id"] for example in EXAMPLES_TO_ADD}.issubset(updated_example_ids):
    raise RuntimeError("Bronze and gold SQL examples were not added.")

print(f"Updated instructions for: {updated_space.title}")
print("Verified explicit text instructions and bronze/gold SQL examples.")
```

```python
from databricks.sdk import WorkspaceClient

SOURCE_SPACE_ID = "01f1a09b07691e458088dc367cd0e81b"
TEST_SPACE_TITLE = "Healthcare Provider Location Data - SDK Create Test"
TEST_SPACE_DESCRIPTION = "Created by exploring.ipynb to validate the Genie create_space API."

w = WorkspaceClient()
spaces = w.genie.list_spaces().spaces or []
existing_space = next(
    (space for space in spaces if space.title == TEST_SPACE_TITLE),
    None,
)

if existing_space:
    created_space = existing_space
    print(f"Using existing test space: {created_space.space_id}")
else:
    source_space = w.genie.get_space(
        space_id=SOURCE_SPACE_ID,
        include_serialized_space=True,
    )

    if not source_space.serialized_space:
        raise RuntimeError("Source Genie Space does not contain serialized_space.")

    created_space = w.genie.create_space(
        warehouse_id=source_space.warehouse_id,
        serialized_space=source_space.serialized_space,
        title=TEST_SPACE_TITLE,
        description=TEST_SPACE_DESCRIPTION,
    )
    print(f"Created test space: {created_space.space_id}")

verified_space = w.genie.get_space(
    space_id=created_space.space_id,
    include_serialized_space=True,
)

if verified_space.title != TEST_SPACE_TITLE:
    raise RuntimeError("Created space title did not persist.")

if not verified_space.serialized_space:
    raise RuntimeError("Created space does not contain serialized_space.")

print(f"Verified space: {verified_space.title}")
print(f"Warehouse: {verified_space.warehouse_id}")
```
