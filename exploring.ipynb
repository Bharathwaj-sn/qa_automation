```python
from databricks.sdk import WorkspaceClient

w = WorkspaceClient()
```

```python
# discover catalogs
catalogs = w.catalogs.list()

for catalog in catalogs:
    print(catalog.name)
```

```python
catalog_name = "dev_adls_lakehouse"
```

```python
schemas = w.schemas.list(catalog_name=catalog_name)

for schema in schemas:
    print(schema.name)
```

```python
schema_name = "silver"
```

```python
tables = w.tables.list(
    catalog_name=catalog_name,
    schema_name=schema_name
)

for table in tables:
    print(table.name)
```

```python
volumes = w.volumes.list(
    catalog_name=catalog_name,
    schema_name=schema_name
)

for volume in volumes:
    print(volume.name)
```

```python
table_name = "aetnanj_demographic"
```

```python
table = w.tables.get(
    full_name=f"{catalog_name}.{schema_name}.{table_name}"
)

print(table)
```

```python
print("Name:", table.name)
print("Catalog:", table.catalog_name)
print("Schema:", table.schema_name)
print("Type:", table.table_type)
print("Data Source Format:", table.data_source_format)
print("Comment:", table.comment)
print("Storage Location:", table.storage_location)

print("\nColumns:")

for column in table.columns or []:
    print(
        f"  {column.name} | "
        f"{column.type_name} | "
        f"{column.type_text} | "
        f"Nullable: {column.nullable} | "
        f"Position: {column.position} | "
        f"Comment: {column.comment}"
    )
```
